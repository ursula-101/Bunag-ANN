# 1.	(10 points) Determine whether the student passes (1) or fails (0) based on:
•	Hours studied
•	Hours of sleep
Given weights w1 = 0.6, w2= 0.4, bias = -3
Step function with a threshold of 1 

Predict the output of the following inputs:
1.	(x1, x2) = (8, 7)
2.	(x1, x2) = (3, 4)


```python
# 1) Perceptron Pass/Fail Prediction

import numpy as np

def step_function(x, threshold=1):
    return 1 if x >= threshold else 0

# Given weights and bias
w1, w2, bias = 0.6, 0.4, -3

def perceptron_pass_fail(x1, x2):
    weighted_sum = (w1 * x1) + (w2 * x2) + bias
    return step_function(weighted_sum)

# Test inputs
inputs = [(8, 7), (3, 4)]
outputs = [perceptron_pass_fail(x1, x2) for x1, x2 in inputs]

for inp, out in zip(inputs, outputs):
    print(f"Input {inp} => Output: {out}")
```

# 2.	(10 points) Logic Gate Simulation. Given the following setup for a perceptron, compute its output and verify whether it acts as an AND gate.
Given weights w1 = 1, w2 = 1, bias = -1.5
Step function with a threshold of 0
Inputs	:
	(0,0) , (0,1), (1,0), (1,1)

```python
# 2) Logic Gate Simulation (Check if AND gate)

def perceptron_and(x1, x2, w1=1, w2=1, bias=-1.5, threshold=0):
    weighted_sum = (w1 * x1) + (w2 * x2) + bias
    return 1 if weighted_sum >= threshold else 0

# Test all input pairs
inputs = [(0,0), (0,1), (1,0), (1,1)]
outputs = [perceptron_and(x1, x2) for x1, x2 in inputs]

for inp, out in zip(inputs, outputs):
    print(f"Input {inp} => Output: {out}")

print("")
print("This perceptron simulates an AND gate.")
```

# 3.	(30 points) Perceptron comparison (One vs All). Given the 3 perceptron, using the same input, compute the output and decided on the predicted class is the WINNER. If a tie is present, compare and get the highest weighted sum. 
Step function with a threshold of 1.0

Inputs = [0.5, -1, 2, 1, 0]
Perceptron A Configuration:
•	Weights: WA = [1.0, -0.5, 0.2, 0.1, 0.0]
•	BiasA: 0.2
Perceptron B
•	Weights: WB = [0.2, 0.2, 0.5, -0.4, 0.3]
•	BiasB: 0.0
Perceptron C
•	Weights: WC = [-0.3, -0.1, 0.4, 0.0, 0.2]
•	BiasC: -0.6


```python
# 3) Perceptron Comparison (One vs All)

def perceptron_output(inputs, weights, bias, threshold=1.0):
    weighted_sum = np.dot(inputs, weights) + bias
    output = 1 if weighted_sum >= threshold else 0
    return output, weighted_sum

# Inputs
inputs = np.array([0.5, -1, 2, 1, 0])

# Perceptron A
WA = np.array([1.0, -0.5, 0.2, 0.1, 0.0])
BiasA = 0.2
outA, sumA = perceptron_output(inputs, WA, BiasA)

# Perceptron B
WB = np.array([0.2, 0.2, 0.5, -0.4, 0.3])
BiasB = 0.0
outB, sumB = perceptron_output(inputs, WB, BiasB)

# Perceptron C
WC = np.array([-0.3, -0.1, 0.4, 0.0, 0.2])
BiasC = -0.6
outC, sumC = perceptron_output(inputs, WC, BiasC)

# Print with 2 decimal places
print(f"Perceptron A => Output: {outA}, Weighted Sum: {sumA:.2f}")
print(f"Perceptron B => Output: {outB}, Weighted Sum: {sumB:.2f}")
print(f"Perceptron C => Output: {outC}, Weighted Sum: {sumC:.2f}")

# Decide the winner
outputs = {'A': (outA, sumA), 'B': (outB, sumB), 'C': (outC, sumC)}

# Step 1: Filter those with output = 1
candidates = {k: v for k, v in outputs.items() if v[0] == 1}

if candidates:
    # Step 2: Compare weighted sums if multiple
    winner = max(candidates, key=lambda k: candidates[k][1])
else:
    winner = "No perceptron activated"

print("")
print(f"Predicted Class Winner: {winner}")
```
